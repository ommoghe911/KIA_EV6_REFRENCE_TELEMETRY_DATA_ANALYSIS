```python
df["Timestamp"] = pd.to_datetime(df["Timestamp"])
```

```python
df["Epoch"] = df["Epoch"].astype("Int64")

#Int data type can handle NaN means missing values ko.
```

 CHECKING OF MISSING VALUES

```python
missing_report = pd.DataFrame({
    "Missing_Count": df.isnull().sum(),
    "Missing_%": (df.isnull().mean() * 100).round(2)
})

missing_report[
    missing_report["Missing_Count"] > 0
].sort_values("Missing_Count", ascending=False)
```

```python
df = df.sort_values(
    ["Session_ID", "Timestamp"]
).reset_index(drop=True)
```

```python
sensor_cols = [
    "TPMS_FL",
    "TPMS_FR",
    "TPMS_RL",
    "TPMS_RR",
    "Battery_Current_A",
    "Battery_SOC_pct",
    "Battery_Voltage_V",
    "Battery_Temp_C",
    "Min_Cell_Voltage_V",
    "Max_Cell_Voltage_V",
    "Available_Discharge_Power_kW",
    "Available_Charge_Power_kW",
    "Vehicle_Speed_kmh",
    "Roll_deg",
    "Heading_deg",
    "Altitude_m",
    "Pitch_deg",
    "Latitude_deg",
    "Longitude_deg",
    "Accel_X_mps2",
    "Accel_Y_mps2",
    "Accel_Z_mps2",
    "Road_Grade_pct"
]
```

```python
for col in sensor_cols:
    df[col] = (
        df.groupby("Session_ID")[col]
        .transform(
            lambda x: x.interpolate(
                method="linear",
                limit=5,
                limit_area="inside"
            )
        )
    )
```

```python
df[sensor_cols].isnull().sum()
```

                                          PHASE 2 DATA CLEANING

                FOR CLEANING nd HANDLING MISSING DATA FROM VOLTAGE , CURRENT AND POWER 
                

```python
for col in ["Battery_Voltage_V", "Battery_Current_A"]:
    df[col] = (
        df.groupby("Session_ID")[col]
        .transform(
            lambda x: x.interpolate(
                method="linear",
                limit=5,
                limit_area="inside"
            )
        )
    )
```

```python
df[["Battery_Voltage_V", "Battery_Current_A"]].isna().sum()
```

```python
df[["Battery_Voltage_V", "Battery_Current_A"]]
```

```python
df[["Battery_Voltage_V", "Battery_Current_A"]].isna().mean() * 100
```

```python
df["Battery_Power_Calculated_kW"] = (
    df["Battery_Voltage_V"] *
    df["Battery_Current_A"]
) / 1000
```

                   CELL VOLTAGE SPREAD 
            
             Cell Voltage Spread = Max Cell Voltage − Min Cell Voltage

```python
df["Cal_cell_voltage_spread"] = (df["Max_Cell_Voltage_V"] - df["Min_Cell_Voltage_V"] )
```
